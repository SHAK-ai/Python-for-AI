```python
class Shape:
    def area():
        return 0

class Rectangle(Shape):
    def area(width, height):
        return f"AREA iS: ", width * height

class Circle(Shape):
    def area( radius):
        return f"AREA is: " ,2 * 3.14 * radius ** 2

lists = [Shape.area(), Rectangle.area(29,10), Circle.area(33)]

for i in lists:
    print(i) 

```
