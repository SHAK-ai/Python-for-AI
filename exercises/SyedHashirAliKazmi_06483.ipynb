**2.** Write a class `Product` so that `Product("pen", 50, 10)` has a `name`, a `price` and a `stock` (stock defaults to `0`). Give it three methods:

- `restock(qty)` adds `qty` to the stock.
- `sell(qty)` takes `qty` away and returns `True`. If there isn't enough stock, it changes **nothing** and returns `False`.
- `value()` returns `price × stock`.

Test your code with:
```python
pen = Product("pen", 50, 10)
print(pen.sell(4), pen.stock)
print(pen.sell(15), pen.stock)
pen.restock(4)
print(pen.value())
```
Expected output:
```
True 6
False 6
500
```

````python
# Your practice space — question 1
#SYED HASHIR ALI KAZMI 
#FROM YOUR TURN Q1


def top_items(n, *sales):
    sorted_sales = sorted(sales,key = lambda s:s[1], reverse=True)
    return [item[0] for item in sorted_sales[:n]]

print(top_items(2, ("pen", 50), ("book", 300), ("bag", 120)))
print(top_items(5, ("pen", 50), ("book", 300)))
print(top_items(3))
````

````python
# Your practice space — question 2
#SYED HASHIR ALI KAZMI 
#FROM YOUR TURN Q2

class product:
    def __init__(self, name, price, stock = 0):
        self.name = name
        self.price = price
        self.stock = stock
    def restock(self, qty):
        self.stock += qty

    def sell(self, qty):
        if qty <= self.stock:
            self.stock -= qty
            return True
        return False
    def value(self):
        return self.price * self.stock

pen = product("PEN", 50, 10)
#print("STOCK OF PEN rn:", pen.stock)

print(pen.sell(4), pen.stock)

print(pen.sell(15), pen.stock)
pen.restock(4)

print(pen.value())
````
